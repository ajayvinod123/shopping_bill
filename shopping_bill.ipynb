```python

def shopping_bill():
    products = []

    while True:
        try:
            n = int(input("Enter number of products: "))
            if n > 0:
                break
            else:
                print("Enter a positive number")
        except ValueError:
            print("Please enter a whole number")

    for i in range(n):
        print("Product", i + 1)
        name = input("Enter product name: ")

        while True:
            try:
                price = int(input("Enter unit price: "))
                if price > 0:
                    break
                else:
                    print("Price must be greater than zero!")
            except ValueError:
                print("Enter a valid price")

        while True:
            try:
                qty = int(input("Enter quantity: "))
                if qty > 0:
                    break
                else:
                    print("Quantity must be greater than zero!")
            except ValueError:
                print("Enter a whole number")

        subtotal = price * qty

        products.append({
            "name": name,
            "price": price,
            "qty": qty,
            "subtotal": subtotal
        })

    total = sum(p["subtotal"] for p in products)

    if total >= 2000:
        discount = total * 10 / 100
    else:
        discount = 0

    after_discount = total - discount
    tax = after_discount * 5 / 100
    final_amount = after_discount + tax

    most_expensive = max(products, key=lambda p: p["price"])

    print("\n========== SHOPPING BILL ==========")

    for p in products:
        print("\nProduct:", p["name"])
        print("Price:", p["price"])
        print("Quantity:", p["qty"])
        print("Subtotal:", p["subtotal"])

    print("\nTotal before discount:", round(total, 2))
    print("Discount:", round(discount, 2))
    print("Tax:", round(tax, 2))
    print("Final Amount:", round(final_amount, 2))
    print("Most Expensive Product:", most_expensive["name"])


shopping_bill()
```
